```python

import numpy as np
import matplotlib.pyplot as plt

# ==========================================
# NOISE-RESILIENT QUANTUM WALK EXPERIMENT
# ==========================================

STEPS = 30
RUNS = 300

# Compare several noise strengths
NOISE_LEVELS = [0.05, 0.15, 0.25, 0.40]

# Small variation between two correlated noise intervals
JITTER = 0.03

rng = np.random.default_rng(42)

# Coin operators
H = np.array([[1, 1], [1, -1]], dtype=complex) / np.sqrt(2)
X = np.array([[0, 1], [1, 0]], dtype=complex)

SIZE = 2 * STEPS + 1
ORIGIN = STEPS


# ==========================================
# QUANTUM WALK
# ==========================================

def coin_operation(psi):
    return psi @ H.T


def conditional_shift(psi):
    out = np.zeros_like(psi)

    # Coin state 0 moves left
    out[:-1, 0] = psi[1:, 0]

    # Coin state 1 moves right
    out[1:, 1] = psi[:-1, 1]

    return out


def walk_step(psi):
    # Coin followed by conditional position shift
    return conditional_shift(coin_operation(psi))


def phase_noise(psi, phi):
    # Apply Rz(phi) to the coin at every position
    out = psi.copy()
    out[:, 0] *= np.exp(-1j * phi / 2)
    out[:, 1] *= np.exp(1j * phi / 2)
    return out


def x_pulse(psi):
    # Pauli-X pulse on the coin
    return psi @ X.T


def probability_distribution(psi):
    return np.sum(np.abs(psi)**2, axis=1)


def fidelity(psi, ideal):
    # Pure-state fidelity with ideal reference
    overlap = np.vdot(ideal, psi)
    return abs(overlap)**2


# ==========================================
# IDEAL REFERENCE EVOLUTION
# ==========================================

initial = np.zeros((SIZE, 2), dtype=complex)
initial[ORIGIN, 0] = 1.0

ideal_history = [initial.copy()]
ideal = initial.copy()

for _ in range(STEPS):
    ideal = walk_step(ideal)
    ideal_history.append(ideal.copy())


# ==========================================
# NOISY AND MITIGATED EVOLUTIONS
# ==========================================

def simulate(noise_strength, jitter, seed):
    local_rng = np.random.default_rng(seed)

    noisy_fid = np.zeros(STEPS + 1)
    mitigated_fid = np.zeros(STEPS + 1)

    noisy_final = np.zeros(SIZE)
    mitigated_final = np.zeros(SIZE)

    for _ in range(RUNS):
        noisy = initial.copy()
        mitigated = initial.copy()

        noisy_fid[0] += 1
        mitigated_fid[0] += 1

        for t in range(1, STEPS + 1):

            # Same intended walk operation in both cases
            noisy = walk_step(noisy)
            mitigated = walk_step(mitigated)

            # Two correlated phase-noise intervals
            phi1 = local_rng.normal(0, noise_strength)
            phi2 = phi1 + local_rng.normal(0, jitter)

            # Baseline: both phase rotations accumulate
            noisy = phase_noise(noisy, phi1)
            noisy = phase_noise(noisy, phi2)

            # Mitigation: echo pulse sequence
            mitigated = phase_noise(mitigated, phi1)
            mitigated = x_pulse(mitigated)
            mitigated = phase_noise(mitigated, phi2)
            mitigated = x_pulse(mitigated)

            # Compare with ideal evolution at this step
            noisy_fid[t] += fidelity(noisy, ideal_history[t])
            mitigated_fid[t] += fidelity(
                mitigated, ideal_history[t]
            )

        noisy_final += probability_distribution(noisy)
        mitigated_final += probability_distribution(mitigated)

    noisy_fid /= RUNS
    mitigated_fid /= RUNS
    noisy_final /= RUNS
    mitigated_final /= RUNS

    return noisy_fid, mitigated_fid, noisy_final, mitigated_final


# ==========================================
# EXPERIMENT 1: SWEEP NOISE STRENGTH
# ==========================================

results = {}

for i, noise in enumerate(NOISE_LEVELS):
    results[noise] = simulate(noise, JITTER, seed=100 + i)

print("FINAL FIDELITY COMPARISON")
print("Noise strength | Noisy | Mitigated | Difference")

for noise in NOISE_LEVELS:
    noisy_f, mitigated_f, _, _ = results[noise]

    print(
        f"{noise:14.2f} | "
        f"{noisy_f[-1]:.4f} | "
        f"{mitigated_f[-1]:.4f} | "
        f"{mitigated_f[-1] - noisy_f[-1]:+.4f}"
    )


# ==========================================
# GRAPH 1: FIDELITY VS NOISE STRENGTH
# ==========================================

noisy_end = [results[n][0][-1] for n in NOISE_LEVELS]
mitigated_end = [results[n][1][-1] for n in NOISE_LEVELS]

plt.figure(figsize=(8, 5))
plt.plot(NOISE_LEVELS, noisy_end, "o-", label="Noisy walk")
plt.plot(NOISE_LEVELS, mitigated_end, "s-",
         label="Echo-mitigated walk")
plt.xlabel("Phase-noise strength (radians)")
plt.ylabel("Final fidelity")
plt.title("Does Dynamical Decoupling Reduce Noise Effects?")
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)
plt.legend()
plt.show()


# ==========================================
# GRAPH 2: FIDELITY OVER TIME
# ==========================================

reference_noise = 0.25
noisy_f, mitigated_f, noisy_p, mitigated_p = results[reference_noise]

plt.figure(figsize=(8, 5))
plt.plot(range(STEPS + 1), noisy_f, label="Noisy walk")
plt.plot(range(STEPS + 1), mitigated_f,
         label="Echo-mitigated walk")
plt.xlabel("Walk step")
plt.ylabel("Fidelity with ideal state")
plt.title("Fidelity Over Time")
plt.ylim(0, 1.05)
plt.grid(alpha=0.3)
plt.legend()
plt.show()


# ==========================================
# GRAPH 3: POSITION DISTRIBUTION
# ==========================================

positions = np.arange(-STEPS, STEPS + 1)
ideal_p = probability_distribution(ideal_history[-1])

plt.figure(figsize=(10, 5))
plt.plot(positions, ideal_p, "--", label="Ideal walk")
plt.plot(positions, noisy_p, label="Noisy walk")
plt.plot(positions, mitigated_p, label="Echo-mitigated walk")
plt.xlabel("Position")
plt.ylabel("Probability")
plt.title("Position Distribution After 30 Steps")
plt.grid(alpha=0.3)
plt.legend()
plt.show()


# ==========================================
# VALIDATION
# ==========================================

print("\nPROBABILITY CONSERVATION CHECK")
print(f"Ideal:     {ideal_p.sum():.6f}")
print(f"Noisy:     {noisy_p.sum():.6f}")
print(f"Mitigated: {mitigated_p.sum():.6f}")
```
